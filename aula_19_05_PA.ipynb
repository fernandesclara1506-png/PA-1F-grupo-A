Escrever um programa que recebe as notas dos 04 bimestres e imprime a média com o resultado de aprovado, recuperação ou reprovado

regras: ultilizar while;

Nota > = 7 aprovado;

Nota > = 5 recuperação;

Nota < 5 reprovado.

```python
soma = 0
contador = 1

while contador <= 4:
    nota = float(input(f"Digite a nota do {contador}º bimestre: "))
    soma += nota
    contador += 1

media = soma / 4

print(f"Média final: {media}")

if media >= 7:
    print("Você está: Aprovado")
elif media >= 5:
    print("Você está: Recuperação")
else:
    print("Você está: Reprovado")
```

Projeto caixa eletrônico

Você foi contratado para desenvolver um sistema simples de caixa eletrônico.

O sistema deverá funcionar através de um menu interativo utilizando “while”

O programa deve iniciar com um saldo inicial de R$1000,00 e permitir que o usuário escolha diferentes operações até decidir sair do sistema.


```python
saldo = 1000.00
opcao = 0

while opcao != 4:
    print("    CAIXA ELETRÔNICO   ")
    print("opção 1: Consultar saldo")
    print("opção 2: Depositar")
    print("opção 3: Sacar")
    print("opção 4: Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        print(f"Seu saldo atual é: R${saldo}")

    elif opcao == 2:
        deposito = float(input("Digite o valor do depósito: R$"))

        if deposito > 0:
            saldo += deposito
            print("Depósito realizado com sucesso!")
        else:
            print("Valor inválido.")

    elif opcao == 3:
        saque = float(input("Digite o valor do saque: R$"))

        if saque <= saldo:
            saldo -= saque
            print("Saque realizado com sucesso!")
        else:
            print("Saldo insuficiente.")

    elif opcao == 4:
        print("saindo do sistema")

    else:
        print("Opção inválida. Tente mais uma vez!")

print("Estarei a sua disposição!")
```
