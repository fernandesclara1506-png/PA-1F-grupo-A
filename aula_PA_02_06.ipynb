Fazer um jogo de advinhação

o usuário irá digitar um número de 0 a 10 para adivinhar o número secreto

o usuário tera 03 tentativas para acertar o número

o programa deverá imprimir uma mensagem se o palpite do usuário é maior ou menor q o número secreto

o programa deverá imprimir uma mensagem de quente, morno ou frio

Regras:se a diferença do palpite para o número secretor for 3- frio; 2-morno;1- quente.

```python
import random

a = random.randint(0, 10)
tentativas_max = 0


while tentativas_max < 3:
    numero = int(input("Digite um número de 0 a 10: "))

    if numero == a:
        print("Está correto!")
        break
    elif numero > a:
        print("Seu palpite é maior que o número secreto.")
    elif numero < a:
        print("Seu palpite é menor que o número secreto.")


    tirar = abs(numero - a)

    if tirar == 1:
        print("Quente!")

    if tirar == 2:
        print("Morno!")

    if tirar >= 3:
        print("Frio!")

    tentativas_max += 1  #

if numero != a:
    print("Você perdeu!")
    print("O número secreto era:", a)
```
