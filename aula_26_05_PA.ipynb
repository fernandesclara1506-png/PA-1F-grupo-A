jogo inteligente

opção 1 par ou impar
opção 2 positivo
opção 3 maior de 2 numeros
opção 4 sair

```python
print("       MENU       ")
print("")
print("Opção 1 - Par / Impar")
print("Opção 2 - Positivo")
print("Opção 3 - Maior")
print("Opção 4 - Sair")
print("")


while True:
    opcao = int(input("digite a opção desejada: "))

    if opcao == 1:

        numero = int(input("digite um número: "))

        if numero % 2 == 0:
            print(f"O numero {numero} é par!")

        else:
            print(f"O numero {numero} é impar!")

    elif opcao == 2:

        if numero > 0:
            print(f"O número {numero} é positivo.")

        else:
            print(f"O número {numero} é negativo.")
```

3 tentativas, bloqueia

```python
senha_correta = 1506
tentativas = 0

while tentativas < 3:
    senha = input("Digite a senha: ")

    if senha == senha_correta:
      print("Senha correta! Acesso concedido.")

    else:
        tentativas += 1
        print("Senha incorreta, tente mais uma vez!")

if tentativas == 3:
     print("Foram 3 tentativas incorretas! Tente novamente mais tarde...")

```
