```python
candidatos = { 1: "Eder", 2: "Rafael Ortega", 3: "JJ"}

votos = { 1: 0, 2: 0, 3: 0, 4: 0}

tC = 0
tv = 0
vo = 0
vos = 0

print("   URNA ELETRÔNICA   ")

while True:
    print("\n   Cadastro do Eleitor   ")
    nome = input("Digite o nome do eleitor: ")

    while True:
        try:
            idade = int(input("Digite a idade do eleitor: "))

            if idade < 0 or idade > 100:
                print("Idade inválida! Digite um valor verdadeiro.")
                continue

            break

        except ValueError:
            print("Entrada inválida! Digite apenas números inteiros para a idade.")

    tC += 1

    if idade < 16:
        print(f"O eleitor {nome} tem {idade} anos e NÃO pode votar.")

    else:
        if 16 <= idade <= 17:
            print(f"O eleitor {nome} tem {idade} anos e o voto é OPCIONAL.")
            vo += 1
        else:
            print(f"O eleitor {nome} tem {idade} anos e é OBRIGATÓRIO votar.")
            vos += 1

        print("\nCandidatos disponíveis:")
        for num, cand_nome in candidatos.items():
            print(f"[{num}] {cand_nome}")

        print("[4] Voto em Branco")

        while True:
            try:
                voto = int(input("Digite o número do seu candidato (ou 4 para branco): "))

                if voto in votos:
                    votos[voto] += 1
                    tv += 1

                    print("Voto registrado!")
                    break
                else:
                    print("Número inválido! Digite um candidato existente.")

            except ValueError:
                print("Entrada inválida! Digite o número do candidato.")

    continuar = input("\nDeseja cadastrar outro eleitor? (Sim/Não): ")

    if continuar.lower() != "sim":
        break


print("\n       RESULTADO       ")
print(f"Total de eleitores cadastrados: {tC}")
print(f"Total de eleitores que votaram: {tv}")
print(f"Quantidade de votos opcionais: {vo}")
print(f"Quantidade de votos obrigatórios: {vos}")

print("\n--- VOTOS ---")

for num, cand_nome in candidatos.items():
    qtd = votos[num]
    percentual = (qtd / tv * 100) if tv > 0 else 0

    print(f"{cand_nome}: {qtd} voto(s) ({percentual:.2f}%)")


qb = votos[4]
pb = (qb / tv * 100) if tv > 0 else 0

print(f"Votos em Branco: {qb} voto(s) ({pb:.2f}%)")


votos_candidatos = {
    num: votos[num]
    for num in candidatos
}

max_votos = max(votos_candidatos.values())

if max_votos == 0:
    print("Nenhum voto válido foi registrado para os candidatos.")

else:
    vencedores = [
        num
        for num, qtd in votos_candidatos.items()
        if qtd == max_votos
    ]

    if len(vencedores) > 1:
        nomes_empate = ", ".join(
            candidatos[num]
            for num in vencedores)

        print( f"Houve EMPATE entre os seguintes candidatos "
            f"({max_votos} votos cada): {nomes_empate}")

    else:
        vencedor_num = vencedores[0]

        print(
            f"CANDIDATO VENCEDOR: "
            f"{candidatos[vencedor_num]} "
            f"com {max_votos} voto(s)!")
```
